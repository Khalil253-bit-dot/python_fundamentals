```python
# Circle Area Calculator
# This program take the radius of a circle in inches and returns the area of the circle
# The program is designed to output the area in square feet and square inches
import math

while True:
    print("Hello! What is the radius of your circle in inches?")
    radius = input("Please enter a numerical value: ")
    try:
        radius = float(radius) # formatting the input into a float. Str by default
        if radius < 0:
            raise ValueError
        break
    except ValueError: #catch inputs that can't be converted to a number
        print("Hey buddy, that's not possible! Try again.")

area = ((radius**2) * math.pi) # ^ unsupported for type float
if area > 144:
    squareFeet = int(area // 144)
    squareInches = area % 144
    print("That's a nice circle you got there! The area of your circle is: ",squareFeet, "square feet and ",squareInches,"square inches")
else:
    print("That's a nice circle you got there! The area of your circle is: ",area,"square inches")
```
